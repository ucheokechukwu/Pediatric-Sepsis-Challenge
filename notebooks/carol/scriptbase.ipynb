# Pediatric Sepsis Challenge

```python
# Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
# Load the data
clean_data = pd.read_csv('../../SyntheticData_training.csv')
```

```python
# Drop 'studyid_adm'
clean_data.drop(columns=['studyid_adm'], inplace=True)
```

```python
# Change datatype 'inhospital_mortality'
clean_data['inhospital_mortality'] = clean_data['inhospital_mortality'].astype('object')
```

```python
# Remove the Intervention Columns as they as strong predictor not allowed by the challenge rules
# Defining the columns to eliminate from the clean_data df.
columns_to_drop = clean_data.columns[clean_data.columns.str.contains('admitabx_adm_')]
columns_to_drop
```

```python
# Remove the intervention columns
clean_data = clean_data.drop(columns=columns_to_drop)
```

```python
# Remove the second Result Column present in the dataset
clean_data.drop(columns=['lengthadm'], inplace = True)
```

```python
# Cheking for Duplicates
clean_data.duplicated().sum()
```

```python
# Several columns with missing values should be filled and we considered better to calculate avg/median/mode based on age groups.
clean_data['age_group'] = pd.cut(
    clean_data['agecalc_adm'], 
    bins=[0, 2, 5, 11, 23, 35, 47, 59, 65],  # Bin edges
    labels=[
        'Newborn (0-2m)', 
        'Infant (3-5m)', 
        'Baby (6-11m)', 
        'Toddler (12-23m)', 
        'Preschool-2 (24-35m)', 
        'Preschool-3 (36-47m)', 
        'Preschool-4 (48-59m)', 
        '5 years (60m)'
    ],
    right=True  # Include the right edge in each interval
)
```

##### Height or length (in cm)

```python
# Calculate the median height for each age group
median_height_by_age_group = clean_data.groupby('age_group')['height_cm_adm'].median()

# Fill missing height values using the median height of the respective age group
def fill_height(row):
    if pd.isnull(row['height_cm_adm']):
        return median_height_by_age_group[row['age_group']]
    return row['height_cm_adm']

clean_data['height_cm_adm'] = clean_data.apply(fill_height, axis=1)
```

##### Weight (in kg)


```python
# Calculate the median weight for each age group
median_height_by_age_group = clean_data.groupby('age_group')['weight_kg_adm'].median()

# Fill missing height values using the median height of the respective age group
def fill_weight(row):
    if pd.isnull(row['weight_kg_adm']):
        return median_height_by_age_group[row['age_group']]
    return row['weight_kg_adm']

clean_data['weight_kg_adm'] = clean_data.apply(fill_weight, axis=1)
```

##### MUAC (in mm)

```python
# Calculate the median MUAC for each age group
median_height_by_age_group = clean_data.groupby('age_group')['muac_mm_adm'].median()

# Fill missing height values using the median height of the respective age group
def fill_muac_mm_adm(row):
    if pd.isnull(row['muac_mm_adm']):
        return median_height_by_age_group[row['age_group']]
    return row['muac_mm_adm']

clean_data['muac_mm_adm'] = clean_data.apply(fill_muac_mm_adm, axis=1)
```

##### Heart rate

```python
# Calculate the median hr_bpm_adm for each age group
median_height_by_age_group = clean_data.groupby('age_group')['hr_bpm_adm'].median()

# Fill missing height values using the median height of the respective age group
def fill_hr_bpm_adm(row):
    if pd.isnull(row['hr_bpm_adm']):
        return median_height_by_age_group[row['age_group']]
    return row['hr_bpm_adm']

clean_data['hr_bpm_adm'] = clean_data.apply(fill_hr_bpm_adm, axis=1)
```

##### Respiratory Rate


```python
# Calculate the median rr_brpm_app_adm for each age group
median_height_by_age_group = clean_data.groupby('age_group')['rr_brpm_app_adm'].median()

# Fill missing height values using the median height of the respective age group
def fill_rr_brpm_app_adm(row):
    if pd.isnull(row['rr_brpm_app_adm']):
        return median_height_by_age_group[row['age_group']]
    return row['rr_brpm_app_adm']

clean_data['rr_brpm_app_adm'] = clean_data.apply(fill_rr_brpm_app_adm, axis=1)
```

##### Systolic Blood Pressure

```python
# Calculate the median sysbp_mmhg_adm for each age group
median_height_by_age_group = clean_data.groupby('age_group')['sysbp_mmhg_adm'].median()

# Fill missing height values using the median height of the respective age group
def fill_sysbp_mmhg_adm(row):
    if pd.isnull(row['sysbp_mmhg_adm']):
        return median_height_by_age_group[row['age_group']]
    return row['sysbp_mmhg_adm']

clean_data['sysbp_mmhg_adm'] = clean_data.apply(fill_sysbp_mmhg_adm, axis=1)
```

##### Diastolic blood pressure

```python
# Calculate the median diasbp_mmhg_adm for each age group
median_height_by_age_group = clean_data.groupby('age_group')['diasbp_mmhg_adm'].median()

# Fill missing height values using the median height of the respective age group
def fill_diasbp_mmhg_adm(row):
    if pd.isnull(row['diasbp_mmhg_adm']):
        return median_height_by_age_group[row['age_group']]
    return row['diasbp_mmhg_adm']

clean_data['diasbp_mmhg_adm'] = clean_data.apply(fill_diasbp_mmhg_adm, axis=1)
```

#### Oxygen saturation (measure 2)

```python
# Calculate the mean of the spo2site2_pc_oxi_adm column
mean_value = clean_data['spo2site2_pc_oxi_adm'].mean()

# Replace missing values in the temperature column with the average
clean_data['spo2site2_pc_oxi_adm'].fillna(mean_value, inplace=True)
```

##### Columns to eliminate

```python
# List of columns to drop because of high number of NaN and low statistical relevancy.
columns_to_drop_2 = [
    'spo2other_adm', 
    'vaccmeaslessource_adm', 
    'vaccpneumocsource_adm', 
    'vaccdptsource_adm', 
    'nonexclbreastfed_adm', 
    'lactate2_mmolpl_adm'
]

# Drop the specified columns
clean_data = clean_data.drop(columns=columns_to_drop_2)
```

#### Several categorical columns needs to be filled with the Mode:

```python
# List of columns to fill with mode
columns_to_fill = ['temp_c_adm', 'spo2site1_pc_oxi_adm', 'spo2onoxy_adm', 'respdistress_adm', 'caprefill_adm', 'bcseye_adm', 'vaccdpt_adm', 'bcsverbal_adm', 'vaccmeasles_adm', 'priorweekantimal_adm', 'priorhosp_adm', 'prioryearwheeze_adm', 'exclbreastfed_adm', 'totalbreastfed_adm', 'deliveryloc_adm', 'travelmethod_adm', 'traveldist_adm', 'badhealthduration_adm', 'caregiver_adm_new', 'caregiverage_adm', 'momalive_adm', 'momagefirstpregknown_adm', 'momhiv_adm', 'householdsize_adm', 'alivechildren_adm', 'deadchildren_adm', 'watersource_adm', 'waterpure_adm', 'cookloc_adm', 'lightfuel_adm', 'bednet_adm', 'hivstatus_adm', 'malariastatuspos_adm', 'momageknown_adm', 'momage_adm', 'hctpretransfusion_adm', 'lactate_mmolpl_adm', 'glucose_mmolpl_adm', 'sqi1_perc_oxi_adm', 'sqi2_perc_oxi_adm']

# Fill each column with its mode
for column in columns_to_fill:
    if column in clean_data.columns:
        mode_value = clean_data[column].mode()[0]  
        clean_data[column].fillna(mode_value, inplace=True)
```

##### Hematocrit (%)

```python
# Calculate the mean hematocrit_gpdl_adm for each age group
median_height_by_age_group = clean_data.groupby('age_group')['hematocrit_gpdl_adm'].mean()

# Fill missing hematocrit_gpdl_adm values using the median for the respective age group
def fill_hematocrit_gpdl_adm(row):
    if pd.isnull(row['hematocrit_gpdl_adm']):
        return median_height_by_age_group[row['age_group']]
    return row['hematocrit_gpdl_adm']

clean_data['hematocrit_gpdl_adm'] = clean_data.apply(fill_hematocrit_gpdl_adm, axis=1)
```

#### momagefirstpreg_adm

```python
# Column to fill
column_name = 'momagefirstpreg_adm'

# Calculate category distribution (excluding NaN)
category_counts = clean_data[column_name].value_counts(normalize=True)

# Determine the number of NaN values
num_nan = clean_data[column_name].isna().sum()

# Calculate the number of NaN values to distribute per category
nan_distribution = (category_counts * num_nan).round().astype(int)  # Convert proportions to integer counts

# Create a list with the proportional categories
values_to_fill = np.hstack([
    [category] * count for category, count in nan_distribution.items()
])

# Shuffle the values to avoid ordered bias (optional)
np.random.shuffle(values_to_fill)

# Fill the NaN values in the column
clean_data.loc[clean_data[column_name].isna(), column_name] = values_to_fill
```
