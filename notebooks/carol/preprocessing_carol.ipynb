### Column Encoding

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, OrdinalEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

def prepare_encoding_pipeline(df):
    # Identify column types automatically
    numeric_columns = df.select_dtypes(include=['int64', 'float64']).columns
    categorical_columns = df.select_dtypes(include=['object', 'category']).columns
    
    # Preprocessing steps for different column types
    preprocessor = ColumnTransformer(
        transformers=[
            # Numeric columns: impute missing values and scale
            ('num', Pipeline([
                ('imputer', SimpleImputer(strategy='median')),
                ('scaler', StandardScaler())
            ]), numeric_columns),
            
            # Categorical columns: one-hot encoding
            ('cat', Pipeline([
                ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
                ('onehot', OneHotEncoder(handle_unknown='ignore'))
            ]), categorical_columns)
        ])
    
    return preprocessor
```

```python
# Function
def apply_encoding_pipeline(df):
    # Create the preprocessing pipeline
    preprocessor = prepare_encoding_pipeline(df)
    
    # Fit and transform the data
    # This returns a numpy array - you might want to convert back to DataFrame
    X_transformed = preprocessor.fit_transform(df)
    
    # Get feature names
    onehot_columns = preprocessor.named_transformers_['cat']['onehot'].get_feature_names_out(df.select_dtypes(include=['object', 'category']).columns)
    numeric_columns = df.select_dtypes(include=['int64', 'float64']).columns
    
    # Combine column names
    transformed_columns = list(numeric_columns) + list(onehot_columns)
    
    # Convert back to DataFrame
    X_encoded = pd.DataFrame(
        X_transformed, 
        columns=transformed_columns, 
        index=df.index
    )
    
    return X_encoded
```
