````python
from sklearn.decomposition import PCA
import numpy as np
import pandas as pd

# will be loaded into the helper_functions.feature_transformer.py file

class PCAReducer:
    def __init__(self, pca_groups):
        """
        Initialize the PCAReducer with predefined feature groups.

        :param pca_groups: Dictionary where keys are group names, and values are sets of column names.
        """
        self.pca_groups = {}
        for k, v in pca_groups.items():
            self.pca_groups[k] = list(v)
        self.pca_models = {}

    def fit(self, X_train):
        """
        Fit PCA models for each feature group using training data.

        :param X_train: DataFrame containing the training data.
        """
        for key, features in self.pca_groups.items():
            if not all(f in X_train.columns for f in features):
                raise ValueError(f"Some features in '{key}' are missing from X_train.")

            pca = PCA(n_components=1)
            self.pca_models[key] = pca.fit(X_train[features])

    def transform(self, X):
        """
        Transform data using previously fitted PCA models.

        :param X: DataFrame containing the test (or validation) data.
        :return: DataFrame with reduced features, one column per group.
        """
        transformed_data = {}

        for key, features in self.pca_groups.items():
            if key not in self.pca_models:
                raise ValueError(f"PCA model for group '{key}' was not fitted.")

            if not all(f in X.columns for f in features):
                raise ValueError(f"Some features in '{key}' are missing from X.")

            transformed_data[key] = self.pca_models[key].transform(X[features]).flatten()

        return pd.DataFrame(transformed_data, index=X.index)

    def fit_transform(self, X):
        """
        Combines fit and transform methods.
        """
        self.fit(X)
        return self.transform(X)
````

````python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.model_selection import KFold

def train_model(X, y):

    kf = KFold(n_splits=5, shuffle=True, random_state=42)  # 5-fold CV

    # Initialize lists to store metrics from each fold
    precision_list = []
    recall_list = []
    f1_list = []
    support_list = []  # Store support from each fold
    class_1_precision_list = []
    class_1_recall_list = []
    class_1_f1_list = []

    # Store feature importances (averaged across folds)
    feature_importance_list = []

    for _, (train_idx, val_idx) in enumerate(kf.split(X)): 

        # Train-test split within this fold
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        # Train the logistic regression model
        model = LogisticRegression(class_weight='balanced', solver='liblinear')
        model.fit(X_train, y_train)

        # Predictions
        y_pred = model.predict(X_val)

        # Get classification report metrics (you can extract precision, recall, f1-score, etc.)
        report = classification_report(y_val, y_pred, output_dict=True)

        # Store metrics for averaging
        precision_list.append(report['weighted avg']['precision'])
        recall_list.append(report['weighted avg']['recall'])
        f1_list.append(report['weighted avg']['f1-score'])
        support_list.append(report['weighted avg']['support'])

        # Store Class 1's specific metrics
        class_1_precision_list.append(report['1']['precision'])
        class_1_recall_list.append(report['1']['recall'])
        class_1_f1_list.append(report['1']['f1-score'])

        # Store feature importance
        feature_importance_list.append(model.coef_[0])

    # Calculate average metrics across all folds
    avg_precision = np.mean(precision_list)
    avg_recall = np.mean(recall_list)
    avg_f1 = np.mean(f1_list)

    avg_class_1_precision = np.mean(class_1_precision_list)
    avg_class_1_recall = np.mean(class_1_recall_list)
    avg_class_1_f1 = np.mean(class_1_f1_list)

    # Print average classification report
    print(f"Average Precision (weighted): {avg_precision:.4f}")
    print(f"Average Recall (weighted): {avg_recall:.4f}")
    print(f"Average F1-Score (weighted): {avg_f1:.4f}")

    print(f"\nAverage Class 1 Precision: {avg_class_1_precision:.4f}")
    print(f"Average Class 1 Recall: {avg_class_1_recall:.4f}")
    print(f"Average Class 1 F1-Score: {avg_class_1_f1:.4f}")

    # Compute average feature importance
    avg_feature_importance = np.mean(feature_importance_list, axis=0)

    # Convert to DataFrame
    feature_importance_df = pd.DataFrame({'Feature': X.columns, 'Coefficient': avg_feature_importance})
    feature_importance_df = feature_importance_df.sort_values(by='Coefficient', ascending=False)

    return feature_importance_df

train_model(X_df[golden_features], y)
print("\n\n")
feature_importance = train_model(X_df, y)
print("\n\n")
feature_importance = train_model(X_pca, y)
````

# Finding the Optimum Combination of Dimensions

````python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.model_selection import KFold

def train_model_grid(X, y, return_feature_importance_df=False):

    kf = KFold(n_splits=5, shuffle=True, random_state=42)  # 5-fold CV

    # Initialize lists to store metrics from each fold
    precision_list = []
    recall_list = []
    f1_list = []
    support_list = []  # Store support from each fold
    class_1_precision_list = []
    class_1_recall_list = []
    class_1_f1_list = []

    # Store feature importances (averaged across folds)
    feature_importance_list = []

    for _, (train_idx, val_idx) in enumerate(kf.split(X)): 

        # Train-test split within this fold
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        # Train the logistic regression model
        model = LogisticRegression(class_weight='balanced', solver='liblinear')
        model.fit(X_train, y_train)

        # Predictions
        y_pred = model.predict(X_val)

        # Get classification report metrics (you can extract precision, recall, f1-score, etc.)
        report = classification_report(y_val, y_pred, output_dict=True)

        # Store metrics for averaging
        precision_list.append(report['weighted avg']['precision'])
        recall_list.append(report['weighted avg']['recall'])
        f1_list.append(report['weighted avg']['f1-score'])
        support_list.append(report['weighted avg']['support'])

        # Store Class 1's specific metrics
        class_1_precision_list.append(report['1']['precision'])
        class_1_recall_list.append(report['1']['recall'])
        class_1_f1_list.append(report['1']['f1-score'])

        # Store feature importance
        feature_importance_list.append(model.coef_[0])

    # Calculate average metrics across all folds
    avg_precision = np.mean(precision_list)
    avg_recall = np.mean(recall_list)
    avg_f1 = np.mean(f1_list)

    avg_class_1_precision = np.mean(class_1_precision_list)
    avg_class_1_recall = np.mean(class_1_recall_list)
    avg_class_1_f1 = np.mean(class_1_f1_list)

    # Print average classification report
    # print(f"Average Precision (weighted): {avg_precision:.4f}")
    # print(f"Average Recall (weighted): {avg_recall:.4f}")
    # print(f"Average F1-Score (weighted): {avg_f1:.4f}")

    # print(f"\nAverage Class 1 Precision: {avg_class_1_precision:.4f}")
    # print(f"Average Class 1 Recall: {avg_class_1_recall:.4f}")
    # print(f"Average Class 1 F1-Score: {avg_class_1_f1:.4f}")

    # Compute average feature importance
    avg_feature_importance = np.mean(feature_importance_list, axis=0)

    # Convert to DataFrame
    feature_importance_df = pd.DataFrame({'Feature': X.columns, 'Coefficient': avg_feature_importance})
    feature_importance_df = feature_importance_df.sort_values(by='Coefficient', ascending=False)

    if return_feature_importance_df:
        return feature_importance_df
    else:
        return avg_f1, avg_class_1_precision, avg_class_1_recall, avg_class_1_f1
    
def plot_scores(n_vs_score): #f1, class_1_precision, class_1_recall, class_1_f1
    score_df = pd.DataFrame.from_dict(n_vs_score, 
                                      orient='index', 
                                      columns=['F1-score', 'Class 1 Precision', 'Class 1 Recall', 'Class 1 F1-score'])
    score_df_sorted = score_df.sort_values(by='Class 1 Recall', ascending=False)

    # Plot recall and F1-score vs the combinations (first 10 for clarity)
    _, ax1 = plt.subplots(figsize=(10, 6))
    ax1.plot(range(10), score_df_sorted['Class 1 Recall'][:10], 'o', color='blue', label='Recall (Class 1)')
    ax1.set_xlabel('Combination Index')
    ax1.set_ylabel('Recall (Class 1)', color='blue')
    ax1.set_ylim(0.6, 1.0)
    ax1.tick_params(axis='y', labelcolor='blue')

    ax2 = ax1.twinx()
    ax2.plot(range(10), score_df_sorted['Class 1 F1-score'][:10], 'x', color='green', label='F1-score (Class 1)')
    ax2.set_ylabel('F1-score (Class 1)', color='green')
    ax2.set_ylim(0.1, 0.5)
    ax2.tick_params(axis='y', labelcolor='green')

    plt.title('Scores for Feature Combinations (with Cross-Validation)')
    ax1.grid(True)
    plt.tight_layout()
    plt.show()
    return score_df_sorted
````

````python
import itertools
import pandas as pd
import matplotlib.pyplot as plt
import math
from tqdm.notebook import tqdm
from sklearn.model_selection import KFold
from joblib import Parallel, delayed  # Import joblib for parallelization

# Select top N features based on absolute coefficient values
N = 29
top_N_features = feature_importance.sort_values(by='Abs_Coefficient', ascending=False).head(N).Feature.values

# Dictionary to store results
n_vs_score = {}

def evaluate_combination(combination):
    """Function to evaluate a single feature combination."""
    f1, class_1_precision, class_1_recall, class_1_f1 = train_model_grid(X_df[list(combination)], y)
    return combination, [f1, class_1_precision, class_1_recall, class_1_f1]

# Loop over different feature set sizes 
for k in tqdm(range(6, 15)):
    combinations = list(itertools.combinations(top_N_features, k))
    num_combinations = math.comb(N, k)

    # Parallel processing with joblib
    results = Parallel(n_jobs=-1)(delayed(evaluate_combination)(comb) for comb in combinations)

    # Store results
    for combination, scores in results:
        n_vs_score[combination] = scores

# Sort and display scores
score_df_sorted = plot_scores(n_vs_score)
display(score_df_sorted[:5])
display(score_df_sorted.sort_values(by="F1-score", ascending=False)[:5])
````
