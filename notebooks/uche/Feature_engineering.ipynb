# Feature engineering 
- totalbreastfed_adm: How long was your child breastfed in total?
- nonexclbreastfed_adm: Non-exclusive breastfeeding details
- exclbreastfed_adm: For how many months was this child exclusively breastfed (months)?

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
data = pd.read_csv("../../SyntheticData_Training.csv")
data = data.iloc[:, 1:] # remove id column
target = data.columns[-1] # save target column
```

```python
for feature in ["exclbreastfed_adm",
                "nonexclbreastfed_adm",
                "totalbreastfed_adm"]:
    print(feature)
    display(data[feature].info())
    display(data[feature].value_counts())
    print("======================\n")
```

```python
data["breastfed_adm_engnr"] = data["exclbreastfed_adm"].where(
    data["exclbreastfed_adm"] != "never exclusively breastfed",
    data["nonexclbreastfed_adm"]).value_counts()
```
